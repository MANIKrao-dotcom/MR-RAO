```python
#nested dictionary
student ={
    "name":"bhairavi",
    "subjects": {
        "chy" : 99,
        "phy" : 90,
        "bio" : 70
    }
}

print(student["subjects"]["chy"])
print(student["subjects"]["phy"])


print(student.keys())#name and subjects are the keys in the dictionary but it never give the nested keys
print(list(student.keys()))#to type cast  in list
print(len(list(student.keys())))#length of the typecasted dicitnory  into list

```
