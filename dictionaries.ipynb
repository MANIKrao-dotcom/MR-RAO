```python
birthday = {
    "manu" : "63622",
    "siddu" : "98869",
    "yadu" : "89710"
}

name = "yadu"
keys = list(birthday.keys())
i = 0
while i < (len(birthday)):
    if(keys[i] == name):
        print("found name",i)
        break
    else :
        print("searching....")
    i = i + 1   
```

```python
#removing item
# pop() : removes the specified kay and returns the associated value

# del : removes the speified key

#clear : empties the dictionary

karnataka_food = {
    "bangaluru" : "bisi bele bath",
    "mysuru" : "mysuru pak",
    "mangaluru" : "neer dose" 
}
mysuru_food = karnataka_food.pop("mysuru")
print(mysuru_food)# remove the specified key and returns the asssociated value.
  
```

```python
# some common methods  available for dictionary
# keys() : return all the keys in the  dictionary.
karnataka_food = {
    "bangaluru" : "bisi bele bath",
    "mysuru" : "mysuru pak",
    "mangluru" : "neer dose" 
}
print(karnataka_food.keys())
```

```python
# values(): returns all the values inn the  dicionary.
karnataka_food = {
    "bangaluru" : "bisi bele bath",
    "mysuru" : "mysuru pak",
    "mangluru" : "neer dose" 
}

print(karnataka_food.values())
```

```python
# items() : returns key-values pairs as tuples
karnataka_food = {
    "bangaluru" : "bisi bele bath",
    "mysuru" : "mysuru pak",
    "mangluru" : "neer dose" 
}
print(karnataka_food.items())
```

```python
# update() : updates the dictionary with another dictionary or iterable.
karnataka_food = {
    "bangaluru" : "bisi bele bath",
    "mysuru" : "mysuru pak",
    "mangluru" : "neer dose" 
}
new_dishes = {"hubballi" : "girmit"}
karnataka_food.update(new_dishes)
print(karnataka_food)
```

```python
#Unordered: Dictionary keys are not stored in any particular order.
#Mutable: You can change, add, or remove items.
#Keys Must Be Immutable: Keys in a dictionary must be of a data type that is immutable, such as a string, number, or tuple.
#Unique Keys: A dictionary cannot have duplicate keys. If you try to add a duplicate key, the latest value will overwrite the previous one.
```
